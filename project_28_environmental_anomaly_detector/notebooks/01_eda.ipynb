# EDA — Environmental Anomaly Detector

Explore sensor time series, distributions, and correlation analysis
for environmental anomaly detection.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
from scipy import stats
```

## Analysis Steps

1. Load sensor time-series data and inspect temporal coverage
2. Plot distributions of key environmental variables
3. Compute correlation matrix across sensor channels
4. Identify obvious outliers and data quality issues
5. Visualize spatial distribution of monitoring stations

```python
# Load sensor data
sensors = pd.read_csv('../data/raw/sensor_readings.csv', parse_dates=['timestamp'])
print(f'Records: {len(sensors)}')
print(f'Sensors: {sensors.sensor_id.nunique()}')
print(f'Date range: {sensors.timestamp.min()} to {sensors.timestamp.max()}')
sensors.describe()
```

```python
# Distribution plots
env_vars = ['temperature', 'humidity', 'pm25', 'no2', 'ozone']
fig, axes = plt.subplots(1, len(env_vars), figsize=(18, 4))
for ax, col in zip(axes, env_vars):
    sensors[col].dropna().hist(bins=40, ax=ax, edgecolor='k', alpha=0.7)
    ax.set_title(col)
plt.tight_layout()
plt.show()

# Correlation matrix
corr = sensors[env_vars].corr()
plt.figure(figsize=(8, 6))
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Correlation')
plt.xticks(range(len(env_vars)), env_vars, rotation=45)
plt.yticks(range(len(env_vars)), env_vars)
plt.title('Sensor Variable Correlations')
plt.tight_layout()
plt.show()
```
