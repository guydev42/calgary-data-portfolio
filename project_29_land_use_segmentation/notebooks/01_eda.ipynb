# EDA — Land Use Segmentation

Explore terrain features, vegetation indices, and land cover distributions
across regions for unsupervised land-use segmentation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import rasterio
from rasterio.plot import show
```

## Analysis Steps

1. Load terrain and land cover rasters
2. Visualize elevation, slope, and aspect distributions
3. Plot NDVI across the study area
4. Explore existing land cover classes from reference data
5. Check spatial extent and resolution consistency

```python
# Load terrain data
with rasterio.open('../data/raw/dem.tif') as src:
    elevation = src.read(1)
    print(f'DEM shape: {elevation.shape}')
    print(f'Elevation range: {elevation.min():.0f}m to {elevation.max():.0f}m')

# Load land cover shapefile
land_cover = gpd.read_file('../data/raw/land_cover.shp')
print(f'Land cover polygons: {len(land_cover)}')
print(f'Classes: {land_cover.lc_class.value_counts().to_dict()}')
```

```python
# Visualize
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
axes[0].imshow(elevation, cmap='terrain')
axes[0].set_title('Elevation (DEM)')
axes[1].hist(elevation.flatten(), bins=50, edgecolor='k')
axes[1].set_title('Elevation Distribution')
land_cover.plot(column='lc_class', legend=True, ax=axes[2], cmap='Set3')
axes[2].set_title('Land Cover Classes')
plt.tight_layout()
plt.show()
```
