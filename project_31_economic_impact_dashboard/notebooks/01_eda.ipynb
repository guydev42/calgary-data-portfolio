# EDA — Economic Impact Dashboard

Explore economic indicators, environmental variables, and regional trends
to understand environment-economy linkages.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
```

## Analysis Steps

1. Load economic indicators (GDP, employment, crop yields)
2. Load environmental variables (NDVI, precipitation, temperature)
3. Explore temporal trends and seasonal patterns
4. Cross-correlation between environmental and economic series
5. Regional comparison across study areas

```python
# Load economic data
econ = pd.read_csv('../data/raw/economic_indicators.csv', parse_dates=['date'])
print(f'Economic records: {len(econ)}')
print(f'Regions: {econ.region.nunique()}')
print(f'Indicators: {econ.columns.tolist()}')
econ.describe()
```

```python
# Load environmental data
env = pd.read_csv('../data/raw/environmental_variables.csv', parse_dates=['date'])
print(f'Environmental records: {len(env)}')

# Time series plot
fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)
for region in econ.region.unique()[:3]:
    mask = econ.region == region
    axes[0].plot(econ.loc[mask, 'date'], econ.loc[mask, 'gdp_index'], label=region)
axes[0].set_title('GDP Index by Region')
axes[0].legend()

for region in env.region.unique()[:3]:
    mask = env.region == region
    axes[1].plot(env.loc[mask, 'date'], env.loc[mask, 'ndvi_mean'], label=region)
axes[1].set_title('Mean NDVI by Region')
axes[1].legend()
plt.tight_layout()
plt.show()
```
