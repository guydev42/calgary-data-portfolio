# EDA — Site Suitability Recommender

Explore multi-layer geospatial features across regions for site suitability
analysis using Multi-Criteria Decision Analysis (MCDA).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
```

## Analysis Steps

1. Load geospatial layers (soil, water, roads, elevation, land cover)
2. Inspect spatial extent and resolution of each layer
3. Visualize distributions of suitability criteria
4. Check for missing data and spatial coverage gaps
5. Correlation analysis between criteria layers

```python
# Load geospatial layers
soil = gpd.read_file('../data/raw/soil_quality.shp')
water = gpd.read_file('../data/raw/water_bodies.shp')
roads = gpd.read_file('../data/raw/road_network.shp')
sites = gpd.read_file('../data/raw/candidate_sites.shp')

print(f'Candidate sites: {len(sites)}')
print(f'Soil polygons: {len(soil)}')
print(f'Water features: {len(water)}')
print(f'Road segments: {len(roads)}')
```

```python
# Visualize layers
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

soil.plot(column='soil_quality', cmap='YlOrBr', legend=True, ax=axes[0])
axes[0].set_title('Soil Quality')

sites.plot(ax=axes[1], color='red', markersize=10)
water.plot(ax=axes[1], color='blue', alpha=0.5)
axes[1].set_title('Sites & Water Bodies')

roads.plot(ax=axes[2], color='gray', linewidth=0.5)
sites.plot(ax=axes[2], color='red', markersize=10)
axes[2].set_title('Road Network & Sites')

plt.tight_layout()
plt.show()
```
