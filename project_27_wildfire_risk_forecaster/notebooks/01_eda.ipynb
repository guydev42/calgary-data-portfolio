# Exploratory Data Analysis — Wildfire Risk Forecaster

This notebook explores the ODAA GeoTIFF and shapefile data used for wildfire risk forecasting.
We examine NDVI distributions, weather patterns, and spatial coverage of the study area.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import rasterio
from rasterio.plot import show
```

## Analysis Steps

1. Load GeoTIFF raster data and inspect band metadata
2. Visualize NDVI distribution across study area
3. Load and plot weather station data (temperature, humidity, wind)
4. Overlay historical fire perimeters on vegetation map
5. Check for missing data and spatial gaps

```python
# Load wildfire perimeter shapefile
fire_perimeters = gpd.read_file('../data/raw/fire_perimeters.shp')
print(f'Number of fire records: {len(fire_perimeters)}')
print(f'CRS: {fire_perimeters.crs}')
fire_perimeters.head()
```

```python
# Load and visualize NDVI raster
with rasterio.open('../data/raw/ndvi_composite.tif') as src:
    ndvi = src.read(1)
    print(f'Shape: {ndvi.shape}')
    print(f'NDVI range: [{ndvi.min():.3f}, {ndvi.max():.3f}]')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].imshow(ndvi, cmap='RdYlGn')
axes[0].set_title('NDVI Composite')
axes[1].hist(ndvi.flatten(), bins=50, edgecolor='k')
axes[1].set_title('NDVI Distribution')
plt.tight_layout()
plt.show()
```

```python
# Load weather data
weather = pd.read_csv('../data/raw/weather_stations.csv', parse_dates=['date'])
print(f'Weather records: {len(weather)}')
print(f'Date range: {weather.date.min()} to {weather.date.max()}')
weather.describe()
```
