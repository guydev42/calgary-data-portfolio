# EDA — Satellite Image Classifier

Explore GeoTIFF tiles, band statistics, and class distributions
for satellite image classification.

```python
import numpy as np
import matplotlib.pyplot as plt
import rasterio
from PIL import Image
import os
from pathlib import Path
```

## Analysis Steps

1. Inventory GeoTIFF tiles and class labels
2. Inspect band statistics (mean, std, min, max per band)
3. Visualize sample tiles as RGB composites
4. Plot class distribution across the dataset
5. Check for class imbalance and spatial coverage

```python
# Inventory tiles
tile_dir = Path('../data/raw/tiles')
classes = sorted([d.name for d in tile_dir.iterdir() if d.is_dir()])
print(f'Classes found: {classes}')

class_counts = {}
for cls in classes:
    count = len(list((tile_dir / cls).glob('*.tif')))
    class_counts[cls] = count
    print(f'  {cls}: {count} tiles')

# Class distribution
plt.figure(figsize=(10, 5))
plt.bar(class_counts.keys(), class_counts.values(), edgecolor='k')
plt.title('Class Distribution')
plt.ylabel('Number of Tiles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Sample tile inspection
sample_tile = list((tile_dir / classes[0]).glob('*.tif'))[0]
with rasterio.open(sample_tile) as src:
    bands = src.read()
    print(f'Tile shape: {bands.shape} (bands, height, width)')
    print(f'CRS: {src.crs}')
    for i in range(bands.shape[0]):
        print(f'  Band {i+1}: mean={bands[i].mean():.2f}, std={bands[i].std():.2f}')

# Display RGB composite
rgb = np.stack([bands[2], bands[1], bands[0]], axis=-1)
rgb = (rgb - rgb.min()) / (rgb.max() - rgb.min())
plt.figure(figsize=(6, 6))
plt.imshow(rgb)
plt.title(f'Sample tile: {classes[0]}')
plt.axis('off')
plt.show()
```
