# CNN Training — Satellite Image Classifier

Fine-tune ResNet-50 with PyTorch for satellite image classification.
Transfer learning with cosine annealing learning rate scheduler.

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models
from torch.optim.lr_scheduler import CosineAnnealingLR
import matplotlib.pyplot as plt
```

## Training Steps

1. Load pretrained ResNet-50 and modify first conv layer for 4-channel input
2. Replace final FC layer for our number of classes
3. Set up optimizer (AdamW) and cosine annealing scheduler
4. Train with early stopping on validation loss
5. Save best model checkpoint

```python
# Modified ResNet-50 for 4-band satellite input
num_classes = 6

model = models.resnet50(pretrained=True)

# Modify first conv layer: 3 channels -> 4 channels (RGB + NIR)
old_conv = model.conv1
model.conv1 = nn.Conv2d(4, 64, kernel_size=7, stride=2, padding=3, bias=False)
with torch.no_grad():
    model.conv1.weight[:, :3] = old_conv.weight
    model.conv1.weight[:, 3] = old_conv.weight[:, 0]  # Initialize NIR from Red

# Replace classifier head
model.fc = nn.Linear(model.fc.in_features, num_classes)

print(f'Model parameters: {sum(p.numel() for p in model.parameters()):,}')
```

```python
# Training configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

optimizer = optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-4)
scheduler = CosineAnnealingLR(optimizer, T_max=30, eta_min=1e-6)
criterion = nn.CrossEntropyLoss()

# Training loop (placeholder)
num_epochs = 30
print(f'Device: {device}')
print(f'Epochs: {num_epochs}')
print(f'Optimizer: AdamW, lr=1e-4')
print(f'Scheduler: CosineAnnealingLR, T_max={num_epochs}')
print('Training loop ready to execute with DataLoaders...')
```
