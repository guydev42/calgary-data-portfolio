# Exploratory data analysis: industry benchmarks

**Project 26** | Industry benchmark engine

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')

df = pd.read_csv('../data/industry_benchmark.csv')
print(f'Dataset: {df.shape[0]} companies, {df.shape[1]} features')
print(f'Industries: {df["industry"].nunique()}')
df.head()
```

```python
df.describe().round(2)
```

```python
kpis = ['turnover_rate','engagement_score','eNPS','training_hours_per_employee','revenue_per_employee','profit_margin']
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, kpi in zip(axes.flat, kpis):
    sns.boxplot(data=df, x='industry', y=kpi, ax=ax, palette='Set2')
    ax.set_title(kpi.replace('_',' ').title())
    ax.tick_params(axis='x', rotation=45)
plt.suptitle('KPI distributions by industry', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
summary = df.groupby('industry')[kpis].mean().round(3)
print('Industry averages:')
summary
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.heatmap(summary.T, annot=True, fmt='.3f', cmap='YlOrRd', ax=ax)
ax.set_title('Industry KPI heatmap')
plt.tight_layout()
plt.show()
```
