# Benchmark analysis: percentile rankings

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/industry_benchmark.csv')
kpis = ['turnover_rate','engagement_score','eNPS','revenue_per_employee']

def percentile_card(df, company_id, metrics):
    company = df[df['company_id'] == company_id].iloc[0]
    industry = company['industry']
    peers = df[df['industry'] == industry]
    results = []
    for m in metrics:
        val = company[m]
        pct = (peers[m] < val).mean() * 100
        results.append({'metric': m, 'value': round(val, 3), 'percentile': round(pct, 1), 'industry_median': round(peers[m].median(), 3)})
    return pd.DataFrame(results)

card = percentile_card(df, df['company_id'].iloc[0], kpis)
print(f'Company: {df["company_id"].iloc[0]} ({df["industry"].iloc[0]})')
card
```

```python
# Gap to 75th percentile by industry
gaps = []
for ind in df['industry'].unique():
    sub = df[df['industry'] == ind]
    for kpi in kpis:
        p75 = sub[kpi].quantile(0.75)
        gaps.append({'industry': ind, 'kpi': kpi, 'p75': round(p75, 3), 'gap_to_p75': round(p75 - sub[kpi].median(), 3)})
gap_df = pd.DataFrame(gaps)
print('Gap analysis (median to 75th percentile):')
gap_df.pivot(index='industry', columns='kpi', values='gap_to_p75').round(3)
```
